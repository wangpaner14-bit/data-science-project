Cell 1: Import Libraries

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Cell 2: Load Data

```python
data = pd.read_csv('northern_surface_temperature.csv',index_col=0)
```

Cell 3: Transpose Data

```python
data_transposed = data.transpose()
```

Cell 4: Create Diverging Colormap

```python
heat_colormap = sns.diverging_palette(240, 15, s=99, as_cmap=True)
```

Cell 5: Generate Heatmap (Every 5 Years)

```python
plt.figure(figsize=(14, 8))
sns.heatmap(data_transposed, cmap=heat_colormap, center=0)
plt.title("Temperature Changes from 1880 to 2015 (base period 1951-1980)")
plt.savefig("temperature_change.png", dpi=300, bbox_inches='tight')
plt.show()
```
